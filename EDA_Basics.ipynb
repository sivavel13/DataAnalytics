```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1.1 Basic Terminology

```python
# import iris dataset into pandas dataframe
iris_ds = pd.read_csv("D:\Data Analysis\Jupyter Notebook\Iris\iris.csv")
```

```python
iris_ds.head()
```

```python
iris_ds.describe()
```

```python
iris_ds["species"].value_counts()
```

```python
iris_ds.plot(kind = "scatter", x = "sepal_length", y = "sepal_width")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.FacetGrid(iris_ds, hue = "species").map(plt.scatter, "sepal_length", "sepal_width").add_legend()
plt.show()
```
